```python
import pandas as pd
```

```python
df = pd.read_csv('/Users/udaykumarreddy/Desktop/Cloud Cost Analysis and Optimization/Cloud Cost Dataset.csv')
```

```python
df.head()
```

```python
print("Dataset Shape:")
print(df.shape)

print("\nColumn Names:")
print(df.columns)

print("\nData Types:")
print(df.dtypes)

print("\nMissing Values:")
print(df.isnull().sum())

print("\nDuplicate Rows:")
print(df.duplicated().sum())

print("\nDataset Information:")
df.info()
```

```python
# Convert date column to date format
df["date"] = pd.to_datetime(df["date"], format="%m/%d/%Y")

# Check numerical columns
print("Numerical Data Summary:")
print(df[["price", "num_instances", "unblended_cost"]].describe())

# Check for invalid values
print("\nZero or Negative Prices:")
print((df["price"] <= 0).sum())

print("\nZero or Negative Instances:")
print((df["num_instances"] <= 0).sum())

print("\nZero or Negative Costs:")
print((df["unblended_cost"] <= 0).sum())

# Check date range
print("\nDate Range:")
print("Start Date:", df["date"].min())
print("End Date:", df["date"].max())
```

```python
# Display rows with zero instances or zero cost
zero_rows = df[
    (df["num_instances"] == 0) |
    (df["unblended_cost"] == 0)
]

print("Number of rows:", len(zero_rows))

zero_rows.head(10)
```

```python
# Save the cleaned dataset
df.to_csv("cloud_cost_cleaned.csv", index=False)

print("Cleaned dataset saved successfully!")
print("Total rows:", len(df))
print("Total columns:", len(df.columns))
```
